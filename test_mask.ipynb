```python
import torch
from transformer_for_ts import Transformer

# Model parameters
d_model = 113
q = 2
v = 2
h = 2
N = 4
d_ff = 64
dropout = 0

d_input = 113
d_output = 9
```

```python
net = Transformer(d_input, d_model, d_output, q, v, h, N, d_ff, dropout=dropout, compute_value=1).cuda()
```

```python
def count_parameters(model):
    return sum(p.numel() for p in model.parameters() if p.requires_grad)
```

```python
print(f'The model has {count_parameters(net.layers_encoding):,} trainable parameters')
```

```python
print(f'The model has {count_parameters(net.layers_decoding):,} trainable parameters')
```

```python
print(f'The model has {count_parameters(net):,} trainable parameters')
```
